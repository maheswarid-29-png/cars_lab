```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://www.cardekho.com/used-cars+in+chennai/page-2"
```

```python
headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

print("Status Code:", response.status_code)
print("Website URL:", response.url)
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
print(soup.prettify()[:5000])
```

```python
import json
```

```python
json_data = soup.find_all(
    "script",
    type="application/ld+json"
)

data = json.loads(json_data[0].string)

print(data.keys())
```

```python
cars = data["@graph"][1]["itemListElement"]

print("Total cars:", len(cars))
```

```python
car_list = []

for car in cars:
    item = car["item"]

    car_details = {
        "Title": item.get("name"),
        "Price": item.get("offers", {}).get("price"),
        "Manufacturing Year": item.get("vehicleModelDate"),
        "Used in Kilometers": item.get("mileageFromOdometer", {}).get("value"),
        "Model Name": item.get("model"),
        "Address": "Chennai"
    }

    car_list.append(car_details)
```

```python
df = pd.DataFrame(car_list)

print(df)
```
